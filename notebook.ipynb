## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

**Pertanyaan 1: Bagaimana hubungan antara ketepatan waktu pengiriman pesanan dengan tingkat kepuasan pelanggan berdasarkan review score selama periode September 2016 hingga Oktober 2018?**

Keterangan:
  - **Specific**: 	Menganalisis hubungan antara ketepatan waktu pengiriman pesanan (tepat waktu dan terlambat) dengan tingkat kepuasan pelanggan berdasarkan review score.
  - **Measurable**: Diukur menggunakan perbandingan tanggal pengiriman aktual (order_delivered_customer_date) dan estimasi pengiriman (order_estimated_delivery_date), serta rata-rata review_score pada setiap pesanan.
  - **Action-Oriented**: 	Hasil analisis dapat digunakan sebagai dasar evaluasi dan perbaikan proses pengiriman untuk meningkatkan kepuasan pelanggan.
  - **Relevant**: Ketepatan waktu pengiriman merupakan salah satu aspek pengalaman pelanggan yang dapat berkaitan dengan tingkat kepuasan terhadap layanan e-commerce.
  - **Time-bound**: 	Analisis dibatasi pada transaksi dengan tanggal pembelian selama September 2016 hingga Oktober 2018.


**Pertanyaan 2: Kategori produk apa saja yang paling sering ditemukan pada pesanan dengan tingkat kepuasan rendah selama periode September 2016 hingga Oktober 2018?**

Keterangan:
  - **Specific**: 	Mengidentifikasi kategori produk yang sering muncul pada pesanan dengan tingkat kepuasan rendah.
  - **Measurable**: Diukur berdasarkan jumlah pesanan unik dengan rata-rata review_score ≤ 2 dan proporsinya terhadap seluruh pesanan yang mengandung kategori tersebut.
  - **Action-Oriented**: 	Hasil analisis dapat digunakan sebagai dasar bagi manajemen untuk menginvestigasi kategori produk yang memerlukan evaluasi lebih lanjut.
  - **Relevant**: Identifikasi kategori yang sering ditemukan pada pesanan dengan kepuasan rendah dapat membantu manajemen menentukan prioritas evaluasi produk dan proses pemenuhan pesanan.
  - **Time-bound**: 	Analisis dibatasi pada transaksi dengan tanggal pembelian selama September 2016 hingga Oktober 2018.

### Cleaning Data

#### Fixing ... problem

```python
# 1. Mengubah tipe data kolom tanggal pada Orders
order_date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for column in order_date_columns:
    orders_df[column] = pd.to_datetime(
        orders_df[column],
        errors='raise'
    )


# 2. Mengubah tipe data kolom tanggal pada Order Reviews
review_date_columns = [
    'review_creation_date',
    'review_answer_timestamp'
]

for column in review_date_columns:
    order_reviews_df[column] = pd.to_datetime(
        order_reviews_df[column],
        errors='raise'
    )


# 3. Mengubah tipe data tanggal pada Order Items
order_items_df['shipping_limit_date'] = pd.to_datetime(
    order_items_df['shipping_limit_date'],
    errors='raise'
)

print("Konversi tipe data tanggal berhasil dilakukan.")


# 4. Memeriksa tipe data setelah konversi
print("\n=== VALIDASI TIPE DATA ===")

for name, df in {
    'Orders': orders_df,
    'Order Reviews': order_reviews_df,
    'Order Items': order_items_df
}.items():
    print(f"\n{name}:")
    print(df.dtypes)


# 5. Menangani missing values pada komentar review
review_comment_columns = [
    'review_comment_title',
    'review_comment_message'
]

for column in review_comment_columns:
    order_reviews_df[column] = (
        order_reviews_df[column].fillna('')
    )

print("\n=== VALIDASI REVIEW COMMENTS ===")
print(order_reviews_df[review_comment_columns].isnull().sum())


# 6. Mempertahankan missing values pada Orders
print("\n=== VALIDASI MISSING VALUES ORDERS ===")
print(orders_df.isnull().sum())

delivered_missing_date = orders_df[
    (orders_df['order_status'] == 'delivered') &
    (orders_df['order_delivered_customer_date'].isnull())
]

print(
    "\nPesanan delivered tanpa tanggal pengiriman aktual:",
    len(delivered_missing_date)
)


# 7. Mempertahankan produk tanpa kategori
print("\n=== VALIDASI MISSING VALUES PRODUCTS ===")
print(products_df.isnull().sum())

missing_category = products_df[
    products_df['product_category_name'].isnull()
]

print("\nJumlah produk tanpa kategori:", len(missing_category))


# 8. Validasi akhir duplikasi dan jumlah baris
datasets = {
    'Orders': orders_df,
    'Order Reviews': order_reviews_df,
    'Order Items': order_items_df,
    'Products': products_df,
    'Category Translation': category_translation_df
}

print("\n=== VALIDASI AKHIR DATASET ===")

for name, df in datasets.items():
    print(
        f"{name}: "
        f"{df.duplicated().sum()} duplikasi | "
        f"{len(df)} baris"
    )
```

**Insight:** (Opsional)

Berdasarkan proses Data Cleaning, dilakukan konversi tipe data pada delapan kolom tanggal di dataset Orders, Order Reviews, dan Order Items menjadi format datetime agar dapat digunakan dalam analisis berbasis waktu. Missing values pada kolom komentar ulasan ditangani dengan menggantinya menggunakan string kosong tanpa menghapus data review.

Sementara itu, missing values pada dataset Orders tetap dipertahankan karena berkaitan dengan status dan proses pengiriman pesanan. Ditemukan delapan pesanan berstatus delivered yang tidak memiliki tanggal pengiriman aktual sehingga tidak dapat digunakan dalam analisis ketepatan waktu pengiriman. Pada dataset Products, terdapat 610 produk tanpa kategori yang akan dikecualikan khusus dalam analisis berdasarkan kategori produk.

Hasil validasi akhir menunjukkan bahwa tidak ditemukan duplikasi pada kelima dataset. Jumlah baris masing-masing dataset juga tetap dipertahankan, yaitu 99.441 baris Orders, 99.224 baris Order Reviews, 112.650 baris Order Items, 32.951 baris Products, dan 71 baris Category Translation. Dengan demikian, proses Data Cleaning telah selesai tanpa menghilangkan data yang masih relevan untuk analisis.

## Exploratory Data Analysis (EDA)

### Explore ...

```python


# 1. AGREGASI REVIEW PER PESANAN


order_review_agg = (
    order_reviews_df
    .groupby('order_id')
    .agg(
        avg_review_score=('review_score', 'mean'),
        total_reviews=('review_score', 'count')
    )
    .reset_index()
)

print("Jumlah pesanan dengan review:",
      order_review_agg['order_id'].nunique())

print("\nStatistik rata-rata review score:")
print(order_review_agg['avg_review_score'].describe())

print("\nJumlah order_id duplikat:",
      order_review_agg['order_id'].duplicated().sum())



# 2. ANALISIS PERTANYAAN BISNIS 1


q1_df = orders_df.merge(
    order_review_agg,
    on='order_id',
    how='inner'
)

q1_df = q1_df[
    (q1_df['order_status'] == 'delivered') &
    (q1_df['order_delivered_customer_date'].notna()) &
    (q1_df['order_estimated_delivery_date'].notna())
].copy()

q1_df['delivery_status'] = (
    q1_df['order_delivered_customer_date'] <=
    q1_df['order_estimated_delivery_date']
).map({
    True: 'Tepat Waktu',
    False: 'Terlambat'
})

print("\nJumlah pesanan untuk analisis Q1:", len(q1_df))

print("\nDistribusi ketepatan waktu pengiriman:")
print(q1_df['delivery_status'].value_counts())

print("\nRata-rata review score berdasarkan ketepatan waktu:")
print(
    q1_df.groupby('delivery_status')['avg_review_score']
    .mean()
)



# 3. ANALISIS PERTANYAAN BISNIS 2


# Mengambil pesanan dengan kepuasan rendah
low_satisfaction_orders = order_review_agg[
    order_review_agg['avg_review_score'] <= 2
].copy()

# Menghubungkan dengan item produk
category_orders = low_satisfaction_orders.merge(
    order_items_df[['order_id', 'product_id']],
    on='order_id',
    how='inner'
)

# Menghubungkan dengan kategori produk
category_orders = category_orders.merge(
    products_df[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

# Menghilangkan produk tanpa kategori
category_orders = category_orders.dropna(
    subset=['product_category_name']
).copy()

# Menghilangkan kategori berulang dalam pesanan yang sama
category_orders = category_orders.drop_duplicates(
    subset=['order_id', 'product_category_name']
)

# Menentukan pesanan dengan kepuasan rendah
category_orders['is_negative'] = (
    category_orders['avg_review_score'] <= 2
)

# Menghitung metrik seluruh pesanan per kategori
all_category_orders = order_review_agg.merge(
    order_items_df[['order_id', 'product_id']],
    on='order_id',
    how='inner'
)

all_category_orders = all_category_orders.merge(
    products_df[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

all_category_orders = all_category_orders.dropna(
    subset=['product_category_name']
).copy()

all_category_orders = all_category_orders.drop_duplicates(
    subset=['order_id', 'product_category_name']
)

# Menghitung jumlah pesanan dan pesanan negatif
category_summary = (
    all_category_orders
    .assign(is_negative=lambda x: x['avg_review_score'] <= 2)
    .groupby('product_category_name')
    .agg(
        total_orders=('order_id', 'nunique'),
        negative_orders=('is_negative', 'sum')
    )
    .reset_index()
)

# Menghitung persentase pesanan negatif
category_summary['negative_order_percentage'] = (
    category_summary['negative_orders'] /
    category_summary['total_orders'] * 100
)

# Menggabungkan terjemahan kategori
category_summary = category_summary.merge(
    category_translation_df,
    on='product_category_name',
    how='left'
)

# Mengurutkan berdasarkan persentase pesanan negatif
category_summary = category_summary.sort_values(
    'negative_order_percentage',
    ascending=False
)

print("\n10 kategori dengan proporsi pesanan negatif tertinggi:")
print(category_summary.head(10).to_string(index=False))


# 4. PEMERIKSAAN KATEGORI TANPA TERJEMAHAN

untranslated_categories = category_summary[
    category_summary['product_category_name_english'].isna()
]

print("\nJumlah kategori tanpa terjemahan:",
      len(untranslated_categories))

print("\nKategori yang belum memiliki terjemahan:")
print(
    untranslated_categories[
        ['product_category_name', 'total_orders', 'negative_orders']
    ].to_string(index=False)
)


```

## Analisis Lanjutan (Opsional)

```python
#ANALISIS LANJUTAN 1

# Menghitung durasi keterlambatan dalam hari
q1_df['delay_days'] = (
    q1_df['order_delivered_customer_date'] -
    q1_df['order_estimated_delivery_date']
).dt.total_seconds() / (24 * 60 * 60)

# Mengubah durasi keterlambatan menjadi kategori
q1_df['delay_category'] = np.select(
    [
        q1_df['delay_days'] <= 0,
        (q1_df['delay_days'] > 0) & (q1_df['delay_days'] <= 3),
        (q1_df['delay_days'] > 3) & (q1_df['delay_days'] <= 7),
        q1_df['delay_days'] > 7
    ],
    [
        'Tepat Waktu',
        'Terlambat 1-3 Hari',
        'Terlambat 4-7 Hari',
        'Terlambat >7 Hari'
    ],
    default='Tidak Diketahui'
)

# Menghitung rata-rata kepuasan berdasarkan kategori keterlambatan
delay_analysis = (
    q1_df.groupby('delay_category')
    .agg(
        total_orders=('order_id', 'nunique'),
        avg_review_score=('avg_review_score', 'mean'),
        avg_delay_days=('delay_days', 'mean')
    )
    .reindex([
        'Tepat Waktu',
        'Terlambat 1-3 Hari',
        'Terlambat 4-7 Hari',
        'Terlambat >7 Hari'
    ])
    .reset_index()
)

# Menampilkan hasil analisis
print("=== ANALISIS DURASI KETERLAMBATAN ===")
print(delay_analysis.to_string(index=False))


# Membuat visualisasi rata-rata kepuasan berdasarkan durasi keterlambatan

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=delay_analysis,
    x='delay_category',
    y='avg_review_score',
    order=[
        'Tepat Waktu',
        'Terlambat 1-3 Hari',
        'Terlambat 4-7 Hari',
        'Terlambat >7 Hari'
    ]
)

plt.title(
    'Hubungan Durasi Keterlambatan dengan Kepuasan Pelanggan',
    fontsize=13,
    fontweight='bold',
    pad=15
)

plt.xlabel('Durasi Keterlambatan')
plt.ylabel('Rata-rata Review Score')
plt.ylim(0, 5)

# Menampilkan nilai di atas batang
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.2f',
        padding=4
    )

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

```
